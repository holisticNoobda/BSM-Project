```
import os, sys, subprocess
from pathlib import Path

REPO = "https://github.com/holisticNoobda/BSM-Project.git"
PROJ = Path("/content/BSM-Project")
BRANCH = "main"

def sh(cmd, check=False):
    print(f"\n$ {cmd}")
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    if r.stdout:
        print(r.stdout[-5000:])
    if r.stderr:
        print("STDERR:", r.stderr[-5000:])
    print(f"[exit {r.returncode}]")
    if check and r.returncode != 0:
        raise SystemExit(r.returncode)
    return r

if PROJ.exists():
    sh(f"rm -rf {PROJ}")
sh(f"git clone --depth 1 -b {BRANCH} {REPO} {PROJ}")
sh("pip install -q -r /content/BSM-Project/requirements.txt", check=False)
sh("python -c \"import nbformat,nbconvert,torch,torchvision; print('ok', torch.cuda.is_available(), torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'cpu')\"")
sh("ls -la /content/BSM-Project/notebooks")
print("\nBOOTSTRAP DONE")
```
